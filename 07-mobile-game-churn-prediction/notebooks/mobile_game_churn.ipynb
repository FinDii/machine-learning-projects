### Experiment notes — 68

Review the implementation and results for this stage of the experiment.

*English presentation summary; the complete original narrative is preserved in `archive/original_notebook.ipynb`.*

```python
# first process function
def build_user_level_features(df):
    """
    第一层聚合处理：将 (user_id, level_id) 级记录汇总
    从原始序列数据构建用户-关卡级别特征

    参数
    ----
    df : pd.DataFrame
        原始数据，需包含字段：user_id, level_id, f_success, f_duration,
        f_reststep, f_help, wd, is_weekend, min_of_day, f_fail_time

    返回
    ----
    user_feat : pd.DataFrame
        以 (user_id, level_id) 为单位的聚合特征表，输出列名以 l_ 开头
    """
    # for data ， sort
    df_sorted = df.sort_values(['user_id', 'level_id', 'time']).copy()
    
    # based on min_of_day compute ， 1440
    if 'min_sin' not in df_sorted.columns:
        df_sorted['min_sin'] = np.sin(2 * np.pi * df_sorted['min_of_day'] / 1440)
        df_sorted['min_cos'] = np.cos(2 * np.pi * df_sorted['min_of_day'] / 1440)
    # compute
    df_sorted['wd_sin'] = np.sin(2 * np.pi * df_sorted['wd'] / 7)
    df_sorted['wd_cos'] = np.cos(2 * np.pi * df_sorted['wd'] / 7)
    
    # 1.
    level_base = df_sorted.groupby(['user_id', 'level_id']).agg(
        l_attempts=('f_success', 'count'),
        l_success_count=('f_success', 'sum'),
        l_help=('f_help', 'mean'),
        l_min_sin=('min_sin', 'mean'),
        l_min_cos=('min_cos', 'mean'),
        l_wd_sin=('wd_sin', 'mean'),
        l_wd_cos=('wd_cos', 'mean'),
        last_attempt_success=('f_success', 'last'),   # delete
        weekend_attempts=('is_weekend', 'sum')        # attempt
    ).reset_index()
    
    level_base['l_wk_ratio'] = level_base['weekend_attempts'] / level_base['l_attempts']
    level_base['l_fail_count'] = level_base['l_attempts'] - level_base['l_success_count']
    level_base['l_success_rate'] = level_base['l_success_count'] / level_base['l_attempts']
    
    # 2. success
    success_subset = df_sorted[df_sorted['f_success'] == 1]
    if not success_subset.empty:
        level_success = success_subset.groupby(['user_id', 'level_id']).agg(
            l_duration_s=('f_duration', 'mean'),
            l_best_duration_s=('f_duration', 'min'),
            l_best_reststep_s=('f_reststep', 'max')
        ).reset_index()
        level_base = level_base.merge(level_success, on=['user_id', 'level_id'], how='left')
    else:
        level_base['l_duration_s'] = 0
        level_base['l_best_duration_s'] = 0
        level_base['l_best_reststep_s'] = 0
    
    # 3. failure
    fail_subset = df_sorted[df_sorted['f_success'] == 0]
    if not fail_subset.empty:
        level_fail = fail_subset.groupby(['user_id', 'level_id']).agg(
            l_fail_time=('f_fail_time', 'mean')
        ).reset_index()
        level_base = level_base.merge(level_fail, on=['user_id', 'level_id'], how='left')
    else:
        level_base['l_fail_time'] = 0
    
    level_base.fillna(0, inplace=True)
    
    # 4. / encoding value
    # Implementation note
    weekend_subset = df_sorted[df_sorted['is_weekend'] == 1]
    if not weekend_subset.empty:
        weekend_agg = weekend_subset.groupby(['user_id', 'level_id']).agg(
            l_wk_min_sin=('min_sin', 'mean'),
            l_wk_min_cos=('min_cos', 'mean')
        ).reset_index()
        level_base = level_base.merge(weekend_agg, on=['user_id', 'level_id'], how='left')
    else:
        level_base['l_wk_min_sin'] = 0
        level_base['l_wk_min_cos'] = 0
    
    # Implementation note
    nweekend_subset = df_sorted[df_sorted['is_weekend'] == 0]
    if not nweekend_subset.empty:
        nweekend_agg = nweekend_subset.groupby(['user_id', 'level_id']).agg(
            l_nwk_min_sin=('min_sin', 'mean'),
            l_nwk_min_cos=('min_cos', 'mean')
        ).reset_index()
        level_base = level_base.merge(nweekend_agg, on=['user_id', 'level_id'], how='left')
    else:
        level_base['l_nwk_min_sin'] = 0
        level_base['l_nwk_min_cos'] = 0
    
    level_base.fillna(0, inplace=True)
    
    # 5. success feature
    def _calc_post(group):
        group = group.sort_values('time')
        successes = group[group['f_success'] == 1]
        if successes.empty:
            return pd.Series({
                'post_success_attempts': 0,
                'post_success_best_improve': 0,
                'l_first_success_duration': 0
            })
        first = successes.iloc[0]
        post = group[group['time'] > first['time']]
        if post.empty:
            return pd.Series({
                'post_success_attempts': 0,
                'post_success_best_improve': 0,
                'l_first_success_duration': first['f_duration']
            })
        attempts = len(post)
        later_success = post[post['f_success'] == 1]
        if later_success.empty:
            improve = 0
        else:
            best_later = later_success['f_duration'].min()
            improve = max(0, first['f_duration'] - best_later)
        return pd.Series({
            'post_success_attempts': attempts,
            'post_success_best_improve': improve,
            'l_first_success_duration': first['f_duration']
        })
    
    post_features = df_sorted.groupby(['user_id', 'level_id']).apply(_calc_post, include_groups=False).reset_index()
    level_base = level_base.merge(post_features, on=['user_id', 'level_id'], how='left')
    level_base.fillna({'post_success_attempts': 0, 'post_success_best_improve': 0, 'l_first_success_duration': 0}, inplace=True)
    
     # delete
    level_base.drop('last_attempt_success', axis=1, inplace=True, errors='ignore')
    level_base.drop('weekend_attempts', axis=1, inplace=True, errors='ignore')
    
    return level_base
```

```python
# second ：user_id
def build_user_all_features(user_level_df, user_seq_df):
    """
    第二层聚合：从 user_level_df 聚合到用户级别，并补充最终关卡信息及重复挑战倾向特征。

    参数:
    user_level_df : pd.DataFrame
        由 build_user_level_features 生成的 (user_id, level_id) 级特征表
    user_seq_df : pd.DataFrame
        原始序列数据，包含 user_id, level_id, f_success, time 等，用于获取最后关卡信息。

    返回:
    pd.DataFrame : 用户级别特征表，每行一个用户。
    """
    # data
    df_level = user_level_df.copy()
    
    # 1. feature
    user_base = df_level.groupby('user_id').agg(
        total_levels=('level_id', 'count'),
        total_attempts=('l_attempts', 'sum'),
        total_success=('l_success_count', 'sum'),
        median_success_rate=('l_success_rate', 'median')
    ).reset_index()
    
    # success = success / attempt
    user_base['avg_success_rate'] = user_base['total_success'] / user_base['total_attempts']
    user_base['total_fail'] = user_base['total_attempts'] - user_base['total_success']
    
    # 2. average feature
    # function ： average （ weights for 0 or value for NaN
    def weighted_mean(df, value_col, weight_col):
        numer = (df[value_col] * df[weight_col]).sum()
        denom = df[weight_col].sum()
        return numer / denom if denom > 0 else 0
    
    # attempt
    user_weighted_by_attempts = df_level.groupby('user_id').apply(
        lambda g: pd.Series({
            'avg_help': weighted_mean(g, 'l_help', 'l_attempts'),
            'avg_min_sin': weighted_mean(g, 'l_min_sin', 'l_attempts'),
            'avg_min_cos': weighted_mean(g, 'l_min_cos', 'l_attempts'),
            'avg_wd_sin': weighted_mean(g, 'l_wd_sin', 'l_attempts'),
            'avg_wd_cos': weighted_mean(g, 'l_wd_cos', 'l_attempts'),
        # / encoding average （ weights for l_attempts
            'avg_wk_min_sin': weighted_mean(g, 'l_wk_min_sin', 'l_attempts'),
            'avg_wk_min_cos': weighted_mean(g, 'l_wk_min_cos', 'l_attempts'),
            'avg_nwk_min_sin': weighted_mean(g, 'l_nwk_min_sin', 'l_attempts'),
            'avg_nwk_min_cos': weighted_mean(g, 'l_nwk_min_cos', 'l_attempts'),
            'avg_wk_ratio': weighted_mean(g, 'l_wk_ratio', 'l_attempts')
        }), include_groups=False
    ).reset_index()
    
    # compute and encoding value
    user_weighted_by_attempts['min_sin_diff'] = (
        user_weighted_by_attempts['avg_wk_min_sin'] - user_weighted_by_attempts['avg_nwk_min_sin']
    )
    user_weighted_by_attempts['min_cos_diff'] = (
        user_weighted_by_attempts['avg_wk_min_cos'] - user_weighted_by_attempts['avg_nwk_min_cos']
    )
    
    # success （ success record level
    success_mask = df_level['l_success_count'] > 0
    df_success = df_level[success_mask].copy()
    user_weighted_by_success = df_success.groupby('user_id').apply(
        lambda g: pd.Series({
            'avg_best_duration': weighted_mean(g, 'l_best_duration_s', 'l_success_count'),
            'avg_duration_s': weighted_mean(g, 'l_duration_s', 'l_success_count'),
            'avg_best_reststep_s': weighted_mean(g, 'l_best_reststep_s', 'l_success_count'),
        }), include_groups=False
    ).reset_index()
    
    # failure （ failure record level
    fail_mask = df_level['l_fail_count'] > 0
    df_fail = df_level[fail_mask].copy()
    user_weighted_by_fail = df_fail.groupby('user_id').apply(
        lambda g: pd.Series({
            'avg_fail_time': weighted_mean(g, 'l_fail_time', 'l_fail_count'),
        }), include_groups=False
    ).reset_index()
    
    # merge all feature
    user_all = user_base.merge(user_weighted_by_attempts, on='user_id', how='left')
    user_all = user_all.merge(user_weighted_by_success, on='user_id', how='left')
    user_all = user_all.merge(user_weighted_by_fail, on='user_id', how='left')
    
    # missing values
    weighted_cols = [
        'avg_help', 'avg_min_sin', 'avg_min_cos', 'avg_wd_sin', 'avg_wd_cos',
        'avg_wk_min_sin', 'avg_wk_min_cos', 'avg_nwk_min_sin', 'avg_nwk_min_cos',
        'min_sin_diff', 'min_cos_diff', 'avg_wk_ratio',
        'avg_best_duration', 'avg_duration_s', 'avg_best_reststep_s', 'avg_fail_time'
    ]
    user_all[weighted_cols] = user_all[weighted_cols].fillna(0)
    
    # 3. high feature ： final level success
    # from take each user last one record （ time sort
    last_df = user_seq_df.sort_values(['user_id', 'time']).groupby('user_id').last().reset_index()
    last_df = last_df[['user_id', 'level_id', 'f_success']]
    last_df.columns = ['user_id', 'fin_level', 'fin_level_success']
    user_all = user_all.merge(last_df, on='user_id', how='left')
    
    # success ratio
    success_levels = df_level[df_level['l_success_count'] > 0].groupby('user_id')['level_id'].count().reset_index()
    success_levels.columns = ['user_id', 'num_success_levels']
    continued_levels = df_level[df_level['post_success_attempts'] > 0].groupby('user_id')['level_id'].count().reset_index()
    continued_levels.columns = ['user_id', 'num_continued_levels']
    user_all = user_all.merge(success_levels, on='user_id', how='left')
    user_all = user_all.merge(continued_levels, on='user_id', how='left')
    user_all['num_success_levels'] = user_all['num_success_levels'].fillna(0).astype(int)
    user_all['num_continued_levels'] = user_all['num_continued_levels'].fillna(0).astype(int)
    user_all['iscontinued_aftersuccess_ratio'] = user_all['num_continued_levels'] / user_all['num_success_levels']
    user_all['iscontinued_aftersuccess_ratio'] = user_all['iscontinued_aftersuccess_ratio'].fillna(0)
    user_all.drop(['num_success_levels', 'num_continued_levels'], axis=1, inplace=True)
    
    # average success （ success
    df_level['improve_rate'] = 0.0
    mask = (df_level['l_first_success_duration'] > 0) & (df_level['post_success_best_improve'] > 0)
    df_level.loc[mask, 'improve_rate'] = df_level.loc[mask, 'post_success_best_improve'] / df_level.loc[mask, 'l_first_success_duration']
    improve_agg = df_level[df_level['improve_rate'] > 0].groupby('user_id').apply(
        lambda g: weighted_mean(g, 'improve_rate', 'l_success_count') if len(g) > 0 else 0, include_groups=False
    ).reset_index(name='avg_post_success_improve_rate')
    user_all = user_all.merge(improve_agg, on='user_id', how='left')
    user_all['avg_post_success_improve_rate'] = user_all['avg_post_success_improve_rate'].fillna(0)
    
    return user_all
```

```python
# function ： cross feature compute
def add_cross_features(user_level_df, level_meta_df, eps=1e-6):
    """
    为 user_level_df 添加用户-关卡交叉对比特征。

    Parameters:
    ---
    user_level_df : pd.DataFrame
        用户关卡聚合特征，需包含列: level_id, l_best_duration_s, l_attempts, l_fail_time, l_success_rate
    level_meta_df : pd.DataFrame
        关卡全局特征，需包含列: level_id, f_avg_win_duration, f_avg_retrytimes,
        f_fail_time, f_avg_passrate, f_retry_intensity
    eps : float
        避免除零的微小常数，默认 1e-6

    Returns:
    ---
    pd.DataFrame : 添加了交叉特征的 user_level_df
    """
    # data
    df = user_level_df.copy()
    meta = level_meta_df[['level_id', 'f_avg_win_duration', 'f_avg_retrytimes', 
                          'f_fail_time', 'f_avg_passrate', 'f_retry_intensity']].drop_duplicates('level_id')
    
    # merge level feature
    df = df.merge(meta, on='level_id', how='left')
    
    # compute cross feature
    df['l_duration_ratio'] = df['l_best_duration_s'] / (df['f_avg_win_duration'] + eps)
    df['l_attempts_ratio'] = df['l_attempts'] / (df['f_avg_retrytimes'] + 1)
    df['l_failtime_ratio'] = df['l_fail_time'] / (df['f_fail_time'] + eps)
    df['l_success_rate_diff'] = df['l_success_rate'] - df['f_avg_passrate']
    df['l_retry_intensity_ratio'] = ((df['l_attempts'] - 1) / (df['l_best_duration_s'] + 1)) / (df['f_retry_intensity'] + eps)
    
    # process or missing values （ for 0 level
    df = df.replace([np.inf, -np.inf], np.nan)
    # feature NaN 0（ or
    df.fillna(0, inplace=True)
    
    return df
```

```python
def aggregate_user_cross_features_weighted(user_level_df, cal_std=True):
    """
    对用户-关卡交叉特征进行用户级加权平均（权重 = l_attempts）。
    
    Parameters:
    user_level_df : pd.DataFrame
        必须包含列: user_id, l_attempts, 以及以下交叉特征列：
        l_duration_ratio, l_attempts_ratio, l_failtime_ratio,
        l_success_rate_diff, l_retry_intensity_ratio
    
    Returns:
    pd.DataFrame : 用户级别的交叉特征加权平均,当cal_std=True时返回标准差
    """
    cross_cols = ['l_duration_ratio', 'l_attempts_ratio', 'l_failtime_ratio',
                  'l_success_rate_diff', 'l_retry_intensity_ratio']
    
    # Implementation note
    existing_cols = [col for col in cross_cols if col in user_level_df.columns]
    if not existing_cols:
        raise ValueError("user_level_df 中缺少交叉特征列，请先运行 add_cross_features")
    
    # for single level weights ， use success ， l_attempts
    # compute average
    def weighted_mean(group, col):
        numer = (group[col] * group['l_attempts']).sum()
        denom = group['l_attempts'].sum()
        return numer / denom if denom > 0 else 0
    
    result = user_level_df.groupby('user_id').apply(
        lambda g: pd.Series({
            f'{col}_wmean': weighted_mean(g, col) for col in existing_cols
        }), include_groups=False
    ).reset_index()
    
    # compute （ user different level table
    if cal_std:
        std_agg = user_level_df.groupby('user_id')[existing_cols].std().add_suffix('_std').reset_index()
        result = result.merge(std_agg, on='user_id', how='left')
    
    result.fillna(0, inplace=True)
    
    return result
```

```python
def prepare_split_data(user_compare_df, user_all_cross_df, split_df, use_user_agg=True, add_log_features=False):
    """
    为单个划分（train/val/test）准备数据。
    
    参数:
    - user_compare_df: 用户-关卡级别特征（含 l_ 开头特征和交叉特征）
    - user_all_cross_df: 用户级别聚合特征（含 avg_ 等特征），若 use_user_agg=False 可忽略
    - split_df: 当前划分的数据集，必须包含 'user_id' 列，若包含 'label' 列则自动提取
    - use_user_agg: 是否将 user_all_cross_df 回填到有序列用户的每一行
    - add_log_features: 是否添加对数变换特征 (log1p)
    
    返回:
    - X: 特征矩阵（多行）
    - y: 标签 Series（若 split_df 有 label），否则 None
    - user_ids: 对应的 user_id 列表
    """
    # take split user
    users_in_split = set(split_df['user_id'])
    df = user_compare_df[user_compare_df['user_id'].isin(users_in_split)].copy()
    
    # user feature
    if use_user_agg and user_all_cross_df is not None:
        df = df.merge(user_all_cross_df, on='user_id', how='left')
        
    # feature
    if add_log_features:
        # use .get() check ， KeyError
        if 'l_duration_s' in df.columns:
            df['l_duration_s_log'] = np.log1p(df['l_duration_s'])
        if 'l_best_duration_s' in df.columns:
            df['l_best_duration_s_log'] = np.log1p(df['l_best_duration_s'])
        if 'l_fail_time' in df.columns:
            df['l_fail_time_log'] = np.log1p(df['l_fail_time'])
        if 'l_help' in df.columns:
            df['l_help_log'] = np.log1p(df['l_help'])
    
    # label （ split_df label
    if 'label' in split_df.columns:
        label_map = split_df[['user_id', 'label']].drop_duplicates()
        df = df.merge(label_map, on='user_id', how='left')
        y = df['label']
    else:
        y = None

    # build one （ level
    df = df.sort_values(['user_id', 'level_id'])
    
    # take feature （ user_id, level_id, label
    exclude_cols = ['user_id', 'level_id', 'label']
    feature_cols = [c for c in df.columns if c not in exclude_cols]
    X = df[feature_cols]
    
    return X, y, df['user_id'].values
```

```python
# final probability
def user_level_auc(y_true, y_pred, user_ids, agg_func='mean'):
    """
    计算用户级别 AUC（多行数据按用户聚合）
    y_true: 多行标签（与 y_pred 长度相同）
    y_pred: 多行预测概率
    user_ids: 对应的 user_id 列表
    agg_func: 聚合方式，'mean', 'max', 'median' 等
    """
    y_pred = np.asarray(y_pred).ravel() # y_pred one
    y_true = np.asarray(y_true).ravel()
    
    df = pd.DataFrame({'user_id': user_ids, 'true': y_true, 'pred': y_pred})
    user_true = df.groupby('user_id')['true'].first()
    if agg_func == 'mean':
        user_pred = df.groupby('user_id')['pred'].mean()
    elif agg_func == 'max':
        user_pred = df.groupby('user_id')['pred'].max()
    elif agg_func == 'median':
        user_pred = df.groupby('user_id')['pred'].median()
    else:
        raise ValueError("agg_func must be 'mean', 'max', or 'median'")
    return roc_auc_score(user_true, user_pred)
```

```python
# manual function （ model and parameters , best parameters and best AUC， prediction )
def grid_search_model(model, param_grid, X_train, y_train, X_val, y_val,
                      train_user_ids, val_user_ids, use_scale=False):
    """
    在训练集上训练模型，验证集上评估用户级别 AUC，返回最佳参数和最佳 AUC
    use_scale: 是否对特征进行标准化（适用于 SVM、LR 等）
    """
    best_auc = -1
    best_params = None
    keys = list(param_grid.keys())
    values = list(param_grid.values())
    for v in product(*values):
        params = dict(zip(keys, v))
        # model parameters
        model_clone = clone(model)
        model_clone.set_params(**params)
        
        # standardization
        if use_scale:
            scaler = StandardScaler()
            X_train_scaled = scaler.fit_transform(X_train)
            X_val_scaled = scaler.transform(X_val)
        else:
            X_train_scaled = X_train
            X_val_scaled = X_val
        model_clone.fit(X_train_scaled, y_train)
        if hasattr(model_clone, "predict_proba"):
            y_val_pred = model_clone.predict_proba(X_val_scaled)[:, 1]
        else:
            y_val_pred = model_clone.decision_function(X_val_scaled)
            y_val_pred = (y_val_pred - y_val_pred.min()) / (y_val_pred.max() - y_val_pred.min())
        auc = user_level_auc(y_val, y_val_pred, val_user_ids)
        if auc > best_auc:
            best_auc = auc
            best_params = params
    return best_params, best_auc
```

```python
def build_sequences(X, y, user_ids, feature_cols):
    """
    将多行数据转换为按用户的序列。
    注！ X 中的数据已经按 (user_id, level_id) 排序，否则会影响顺序。
    返回:
        sequences: list of np.array, shape each (seq_len, n_features)
        labels: list of int (每个用户的标签)
        user_ids_list: list of user_id (与 sequences 顺序对应)
    """
    df = X.copy()
    df['user_id'] = user_ids
    df['label'] = y
    grouped = df.groupby('user_id')
    sequences = []
    labels = []
    user_ids_list = []
    for uid, group in grouped:
        # sort ， use （ X
        seq = group[feature_cols].values  # (seq_len, n_features)
        sequences.append(seq.astype(np.float32))
        labels.append(group['label'].iloc[0])
        user_ids_list.append(uid)
    return sequences, np.array(labels), np.array(user_ids_list)
```

```python
# define PyTorch Dataset and DataLoader
class SequenceDataset(Dataset):
    def __init__(self, sequences, labels):
        self.sequences = sequences
        self.labels = torch.tensor(labels, dtype=torch.float32)
        self.lengths = [len(seq) for seq in sequences]
    
    def __len__(self):
        return len(self.sequences)
    
    def __getitem__(self, idx):
        seq = torch.tensor(self.sequences[idx], dtype=torch.float32)
        label = self.labels[idx]
        length = self.lengths[idx]
        return seq, label, length

def collate_fn(batch):
    seqs, labels, lengths = zip(*batch)
    max_len = max(lengths)
    padded_seqs = torch.zeros(len(seqs), max_len, seqs[0].size(1))
    for i, seq in enumerate(seqs):
        padded_seqs[i, :len(seq), :] = seq
    return padded_seqs, torch.tensor(labels), torch.tensor(lengths)

train_dataset = SequenceDataset(train_seqs_scaled, train_labels)
val_dataset = SequenceDataset(val_seqs_scaled, val_labels)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, collate_fn=collate_fn)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, collate_fn=collate_fn)
```

```python
import random

def train_sequence_model(model, train_loader, val_loader, epochs, lr, device='cuda',
                         patience=5, early_stop=True):
    """
    训练序列模型（GRU/LSTM等），返回最佳验证集AUC和训练好的模型
    """
    model = model.to(device)
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.BCELoss()
    
    best_auc = 0.0
    best_model_state = None
    patience_counter = 0
    
    for epoch in range(1, epochs + 1):
        # training
        model.train()
        train_loss = 0.0
        for batch in train_loader:
            # collate_fn (padded_seqs, labels, lengths)
            if len(batch) == 3:
                seqs, labels, lengths = batch
            else:
                # Implementation note
                seqs, labels = batch
                lengths = torch.tensor([seq.size(0) for seq in seqs])
            seqs, labels = seqs.to(device), labels.to(device)
            lengths = lengths.cpu()  # pack_padded_sequence CPU tensor
            
            optimizer.zero_grad()
            outputs = model(seqs, lengths)   # note ：forward lengths
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
        
        # validation
        model.eval()
        preds = []
        true_labels = []
        with torch.no_grad():
            for batch in val_loader:
                if len(batch) == 3:
                    seqs, labels, lengths = batch
                else:
                    seqs, labels = batch
                    lengths = torch.tensor([seq.size(0) for seq in seqs])
                seqs, labels = seqs.to(device), labels.to(device)
                outputs = model(seqs, lengths)
                preds.extend(outputs.cpu().numpy())
                true_labels.extend(labels.cpu().numpy())
        
        auc = roc_auc_score(true_labels, preds)
        avg_loss = train_loss / len(train_loader)
        print(f"Epoch {epoch}/{epochs} - Loss: {avg_loss:.4f} - Val AUC: {auc:.4f}")
        
        # save best model
        if auc > best_auc:
            best_auc = auc
            best_model_state = model.state_dict().copy()
            patience_counter = 0
        else:
            patience_counter += 1
        
        if early_stop and patience_counter >= patience:
            print(f"Early stopping at epoch {epoch}")
            break
    
    # load best model
    model.load_state_dict(best_model_state)
    return best_auc, model
```

```python
# define model
# LSTM model
class LSTMClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2, dropout=0.5):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first=True, dropout=dropout)
        self.fc = nn.Linear(hidden_dim, 1)
    
    def forward(self, x, lengths):
        # x: (batch, max_len, input_dim)
        packed = nn.utils.rnn.pack_padded_sequence(x, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, (h_n, _) = self.lstm(packed)
        # h_n: (num_layers, batch, hidden_dim) -> take last one
        out = h_n[-1]  # (batch, hidden_dim)
        return torch.sigmoid(self.fc(out)).squeeze()
```

```python
# GRU model
class GRUClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, num_layers=2, dropout=0.5):
        super().__init__()
        self.gru = nn.GRU(input_dim, hidden_dim, num_layers, batch_first=True, dropout=dropout)
        self.fc = nn.Linear(hidden_dim, 1)
    
    def forward(self, x, lengths):
        packed = nn.utils.rnn.pack_padded_sequence(x, lengths.cpu(), batch_first=True, enforce_sorted=False)
        _, h_n = self.gru(packed)
        out = h_n[-1]
        return torch.sigmoid(self.fc(out)).squeeze()
```

```python
# CNN model （ one +
class CNN1DClassifier(nn.Module):
    def __init__(self, input_dim, hidden_dim=64, kernel_size=3):
        super().__init__()
        self.conv = nn.Conv1d(input_dim, hidden_dim, kernel_size, padding=1)
        self.pool = nn.AdaptiveMaxPool1d(1)
        self.fc = nn.Linear(hidden_dim, 1)
    
    def forward(self, x, lengths):
        # x: (batch, max_len, input_dim) -> for (batch, input_dim, max_len)
        x = x.permute(0, 2, 1)
        x = torch.relu(self.conv(x))
        x = self.pool(x).squeeze(-1)  # (batch, hidden_dim)
        return torch.sigmoid(self.fc(x)).squeeze()
```

```python
# ==================== parameters function ====================
def create_dataloaders(batch_size, train_seqs, train_lbls, val_seqs, val_lbls):
    """根据 batch_size 动态创建 DataLoader"""
    train_dataset = SequenceDataset(train_seqs, train_lbls)
    val_dataset = SequenceDataset(val_seqs, val_lbls)
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, collate_fn=collate_fn)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, collate_fn=collate_fn)
    return train_loader, val_loader

def evaluate_single_model(params, train_seqs, train_lbls, val_seqs, val_lbls, epochs, device='cuda'):
    """
    用给定超参数训练 GRU 并返回验证集 AUC
    params: dict 包含 hidden_dim, num_layers, dropout, lr, batch_size
    """
    # create DataLoader
    train_loader, val_loader = create_dataloaders(
        params['batch_size'], train_seqs, train_lbls, val_seqs, val_lbls
    )
    # create model
    model = GRUClassifier(
        input_dim=input_dim,
        hidden_dim=params['hidden_dim'],
        num_layers=params['num_layers'],
        dropout=params['dropout']
    ).to(device)
    # training best validation AUC
    best_auc, _ = train_sequence_model(
        model, train_loader, val_loader, epochs=epochs, lr=params['lr'], device=device
    )
    return best_auc
```

```python
def build_last_features(seq_df, level_meta_df):
    df = seq_df.copy()
    df['time'] = pd.to_datetime(df['time'])
    df = df.sort_values(['user_id', 'time'])

    # merge level
    df = df.merge(level_meta_df, on='level_id', how='left')

    last = df.groupby('user_id').tail(1).copy()

    feat = last[[
        'user_id', 'f_success', 'f_duration',
        'f_reststep', 'f_help', 'time',
        'f_avg_duration', 'f_avg_passrate',
        'f_avg_win_duration', 'f_avg_retrytimes'
    ]].copy()

    feat = feat.rename(columns={
        'f_success': 'fin_level_success',
        'f_duration': 'last_duration',
        'f_reststep': 'last_reststep',
        'f_help': 'last_help',
        'f_avg_duration': 'last_level_avg_duration',
        'f_avg_passrate': 'last_level_passrate',
        'f_avg_win_duration': 'last_level_win_duration',
        'f_avg_retrytimes': 'last_level_retrytimes'
    })

    end_time = pd.Timestamp('2020-02-04 23:59:59')
    feat['recency_hours'] = (end_time - feat['time']).dt.total_seconds() / 3600
    feat['last_day'] = feat['time'].dt.day
    feat['last_hour'] = feat['time'].dt.hour
    feat['last_min_of_day'] = feat['last_hour'] * 60 + feat['time'].dt.minute

    feat['last_is_day4'] = (feat['last_day'] == 4).astype(int)

    feat['last_fail'] = (feat['fin_level_success'] == 0).astype(int)

    feat['last_fail_on_hard_level'] = (
        (feat['fin_level_success'] == 0) &
        (feat['last_level_passrate'] < 0.45)
    ).astype(int)

    feat['last_success_on_hard_level'] = (
        (feat['fin_level_success'] == 1) &
        (feat['last_level_passrate'] < 0.45)
    ).astype(int)

    feat['last_duration_ratio_to_level'] = (
        feat['last_duration'] / (feat['last_level_avg_duration'] + 1e-6)
    )

    feat = feat.drop(columns=['time', 'fin_level_success'])

    return feat
```

```python
def build_lastn_features(seq_df, level_meta_df, n=10):
    df = seq_df.copy()
    df['time'] = pd.to_datetime(df['time'])
    df = df.sort_values(['user_id', 'time'])
    df = df.merge(level_meta_df, on='level_id', how='left')

    lastn = df.groupby('user_id').tail(n).copy()

    feat = lastn.groupby('user_id').agg(
        **{
            f'last{n}_attempts': ('level_id', 'size'),
            f'last{n}_success_rate': ('f_success', 'mean'),
            f'last{n}_fail_count': ('f_success', lambda x: (x == 0).sum()),
            f'last{n}_avg_duration': ('f_duration', 'mean'),
            f'last{n}_max_duration': ('f_duration', 'max'),
            f'last{n}_avg_reststep': ('f_reststep', 'mean'),
            f'last{n}_help_rate': ('f_help', 'mean'),
            f'last{n}_unique_levels': ('level_id', 'nunique'),
            f'last{n}_max_level': ('level_id', 'max'),
            f'last{n}_min_level_passrate': ('f_avg_passrate', 'min'),
            f'last{n}_avg_level_passrate': ('f_avg_passrate', 'mean'),
            f'last{n}_avg_level_retrytimes': ('f_avg_retrytimes', 'mean'),
        }
    ).reset_index()

    feat[f'last{n}_fail_rate'] = 1 - feat[f'last{n}_success_rate']
    feat[f'last{n}_stuck_ratio'] = 1 - feat[f'last{n}_unique_levels'] / feat[f'last{n}_attempts'].clip(lower=1)

    return feat
```

```python
def build_final_level_stuck_features(seq_df, level_meta_df):
    df = seq_df.copy()
    df['time'] = pd.to_datetime(df['time'])
    df = df.sort_values(['user_id', 'time'])
    df = df.merge(level_meta_df, on='level_id', how='left')

    # each user last level
    last_level = df.groupby('user_id')['level_id'].last().reset_index()
    last_level = last_level.rename(columns={'level_id': 'final_level_id'})

    tmp = df.merge(last_level, on='user_id', how='left')
    final_df = tmp[tmp['level_id'] == tmp['final_level_id']].copy()

    feat = final_df.groupby('user_id').agg(
        final_level_id=('final_level_id', 'last'),
        final_level_attempts=('level_id', 'size'),
        final_level_success_count=('f_success', 'sum'),
        final_level_fail_count=('f_success', lambda x: (x == 0).sum()),
        final_level_success_rate=('f_success', 'mean'),
        final_level_avg_duration=('f_duration', 'mean'),
        final_level_max_duration=('f_duration', 'max'),
        final_level_avg_reststep=('f_reststep', 'mean'),
        final_level_help_rate=('f_help', 'mean'),
        final_level_passrate=('f_avg_passrate', 'last'),
        final_level_avg_retrytimes=('f_avg_retrytimes', 'last')
    ).reset_index()

    feat['final_level_last_failed'] = (
        feat['final_level_success_rate'] < 1
    ).astype(int)

    feat['stuck_on_final_level'] = (
        (feat['final_level_attempts'] >= 3) &
        (feat['final_level_success_rate'] < 1)
    ).astype(int)

    feat['hard_stuck_on_final_level'] = (
        (feat['final_level_attempts'] >= 3) &
        (feat['final_level_passrate'] < 0.45) &
        (feat['final_level_success_rate'] < 1)
    ).astype(int)

    feat['final_fail_ratio'] = (
        feat['final_level_fail_count'] / feat['final_level_attempts'].clip(lower=1)
    )

    feat = feat.drop(columns=['final_level_id'])

    return feat
```

```python
def build_fail_streak_features(seq_df):
    df = seq_df.copy()
    df['time'] = pd.to_datetime(df['time'])
    df = df.sort_values(['user_id', 'time'])

    def streak_features(g):
        success = g['f_success'].values

        # last failure
        last_fail_streak = 0
        for v in success[::-1]:
            if v == 0:
                last_fail_streak += 1
            else:
                break

        # failure
        max_fail_streak = 0
        cur = 0
        for v in success:
            if v == 0:
                cur += 1
                max_fail_streak = max(max_fail_streak, cur)
            else:
                cur = 0

        # last success
        last_success_streak = 0
        for v in success[::-1]:
            if v == 1:
                last_success_streak += 1
            else:
                break

        return pd.Series({
            'last_fail_streak': last_fail_streak,
            'max_fail_streak': max_fail_streak,
            'last_success_streak': last_success_streak
        })

    return df.groupby('user_id').apply(streak_features).reset_index()
```

```python
def build_daily_features(seq_df):
    df = seq_df.copy()
    df['time'] = pd.to_datetime(df['time'])
    df['day'] = df['time'].dt.day

    # each user
    daily = df.groupby(['user_id', 'day']).agg(
        attempts=('level_id', 'size'),
        success_rate=('f_success', 'mean'),
        max_level=('level_id', 'max')
    ).reset_index()

    # for table （ only take
    wide_attempts = daily.pivot(index='user_id', columns='day', values='attempts').fillna(0).astype(int)
    wide_success_rate = daily.pivot(index='user_id', columns='day', values='success_rate').fillna(0)
    wide_max_level = daily.pivot(index='user_id', columns='day', values='max_level').fillna(0).astype(int)

    # one （1,2,3,4
    for d in [1, 2, 3, 4]:
        if d not in wide_attempts.columns:
            wide_attempts[d] = 0
        if d not in wide_success_rate.columns:
            wide_success_rate[d] = 0.0
        if d not in wide_max_level.columns:
            wide_max_level[d] = 0

    # compute feature
    active_day4 = (wide_attempts[4] > 0).astype(int)
    attempts_trend_4_1 = wide_attempts[4] - wide_attempts[1]
    success_rate_trend_4_1 = wide_success_rate[4] - wide_success_rate[1]
    level_progress_4_1 = wide_max_level[4] - wide_max_level[1]

    # attempts>0
    active_days = (wide_attempts > 0).sum(axis=1)
    total_active_days = active_days

    # from 1， to first
    # 1+1， then check 2
    consecutive_active_days = pd.Series(0, index=wide_attempts.index)
    for d in [1, 2, 3, 4]:
        consecutive_active_days += (wide_attempts[d] > 0) & (consecutive_active_days == (d-1))
   
    # one bool matrix ， compute
    active_bool = (wide_attempts[[1,2,3,4]] > 0).astype(int)
    # compute （ to 00
    prefix_prod = active_bool.cumprod(axis=1)
    consecutive_active_days = prefix_prod.sum(axis=1)

    # build results DataFrame
    result = pd.DataFrame({
        'user_id': wide_attempts.index,
        'active_day4': active_day4,
        'attempts_trend_4_1': attempts_trend_4_1,
        'success_rate_trend_4_1': success_rate_trend_4_1,
        'level_progress_4_1': level_progress_4_1,
        'total_active_days': total_active_days,
        'consecutive_active_days': consecutive_active_days
    }).reset_index(drop=True)

    return result
```

```python
# DeLong function （ use Bootstrap method
def delong_roc_test(ground_truth, predictions1, predictions2):
    """
    DeLong test for two correlated ROC curves using bootstrap.
    ground_truth: binary array (0/1)
    predictions1, predictions2: probability arrays
    Returns p-value
    """
    from scipy import stats
    n_bootstrap = 2000
    np.random.seed(42)
    indices = np.arange(len(ground_truth))
    diff_bootstrap = []
    for _ in range(n_bootstrap):
        idx = np.random.choice(indices, size=len(indices), replace=True)
        auc1 = roc_auc_score(ground_truth[idx], predictions1[idx])
        auc2 = roc_auc_score(ground_truth[idx], predictions2[idx])
        diff_bootstrap.append(auc1 - auc2)
    diff_mean = np.mean(diff_bootstrap)
    diff_std = np.std(diff_bootstrap)
    p_value = 2 * (1 - stats.norm.cdf(np.abs(diff_mean) / (diff_std + 1e-8)))
    return p_value
```

```python
# load all model test set prediction results （ user ） merge label
model_predictions = {}
for model_name in ['XGBoost', 'LightGBM', 'CatBoost', 'LR', 'Ensemble']:
    if model_name == 'LR':
        fname = 'LR_top100_test_predictions.csv'
    elif model_name == 'Ensemble':
        fname = 'Ensemble_top100_test_predictions.csv'
    else:
        fname = f'{model_name}_top100_test_predictions.csv'
    try:
        df = pd.read_csv(fname)
        if 'label' not in df.columns:
            if ground_truth_df is not None:
                df = df.merge(ground_truth_df, on='user_id')
        model_predictions[model_name] = (df['label'].values, df['pred_prob'].values)
    except FileNotFoundError:
        print(f"未找到 {fname}，跳过{model_name}")

# compare （ model 'Ensemble' for comparison
base_model = 'Ensemble'
for model_name in model_predictions.keys():
    if model_name == base_model:
        continue
    gt, pred1 = model_predictions[base_model]
    _, pred2 = model_predictions[model_name]
    p_val = delong_roc_test(gt, pred1, pred2)
    print(f"{base_model} vs {model_name}: p-value = {p_val:.4f}")

# plot all model ROC
plt.figure(figsize=(10, 8))
for name, (labels, preds) in model_predictions.items():
    fpr, tpr, _ = roc_curve(labels, preds)
    auc_val = roc_auc_score(labels, preds)
    plt.plot(fpr, tpr, label=f'{name} (AUC={auc_val:.4f})', linewidth=2)
plt.plot([0,1], [0,1], 'k--', label='Random')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curves Comparison on Test Set')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.savefig('all_models_roc.png', dpi=150)
plt.show()
```

```python
# take model table and predicted value
model_names = list(model_predictions.keys())
n = len(model_names)

# take label （ all model one label ， take first model
y_true_user, _ = model_predictions[model_names[0]]

# initialize p-value matrix
p_matrix = pd.DataFrame(np.ones((n, n)), index=model_names, columns=model_names)

# DeLong
print("Performing pairwise DeLong tests...")
total_pairs = n * (n - 1) // 2
pair_count = 0
for i in range(n):
    for j in range(i+1, n):
        pair_count += 1
        print(f"  Testing {model_names[i]} vs {model_names[j]} ({pair_count}/{total_pairs})...")
        # note ：delong_roc_test p-value， parameters for (ground_truth, pred1, pred2)
        p_val = delong_roc_test(y_true_user, model_predictions[model_names[i]][1], model_predictions[model_names[j]][1])
        p_matrix.iloc[i, j] = p_val
        p_matrix.iloc[j, i] = p_val

print("\nPairwise DeLong test p-values:")
print(p_matrix.round(4))

# plot plot
plt.figure(figsize=(8, 6))
sns.heatmap(p_matrix, annot=True, cmap='coolwarm', center=0.05, fmt='.4f', 
            cbar_kws={'label': 'p-value'})
plt.title('DeLong Test p-values (Pairwise Comparison)')
plt.tight_layout()
plt.savefig('delong_pvalues_heatmap.png', dpi=150)
plt.show()
```

```python
from sklearn.metrics import brier_score_loss

def calculate_ece(y_true, y_prob, n_bins=10):
    """计算预期校准误差 (ECE)"""
    # probability split for n_bins
    bin_boundaries = np.linspace(0, 1, n_bins + 1)
    bin_lowers = bin_boundaries[:-1]
    bin_uppers = bin_boundaries[1:]
    
    ece = 0.0
    for bin_lower, bin_upper in zip(bin_lowers, bin_uppers):
        # to sample
        in_bin = (y_prob >= bin_lower) & (y_prob < bin_upper)
        prob_in_bin = y_prob[in_bin]
        true_in_bin = y_true[in_bin]
        
        if len(prob_in_bin) > 0:
            # average prediction probability
            avg_prob = np.mean(prob_in_bin)
            # positive samples ratio
            avg_true = np.mean(true_in_bin)
            # value * weights
            ece += np.abs(avg_prob - avg_true) * len(prob_in_bin) / len(y_prob)
    return ece

# results
calibration_results = []

models = long_df['model'].unique()
for m in models:
    subset = long_df[long_df['model'] == m]
    y_true = subset['label']
    y_prob = subset['pred_prob']
    
    # 1. compute Brier Score ()
    brier = brier_score_loss(y_true, y_prob)
    
    # 2. compute ECE ()
    ece = calculate_ece(y_true, y_prob, n_bins=10)
    
    # 3. ： prediction probability ()
    max_prob = y_prob.max()
    mean_pos_prob = y_prob[y_true == 1].mean()
    mean_neg_prob = y_prob[y_true == 0].mean()
    
    calibration_results.append({
        'Model': m,
        'Brier Score': brier,
        'ECE': ece,
        'Max Predicted Prob': max_prob,
        'Mean Prob (True=1)': mean_pos_prob,
        'Mean Prob (True=0)': mean_neg_prob
    })

# convert for DataFrame
res_df = pd.DataFrame(calibration_results)
print("=== 模型置信度（校准度）量化指标 ===")
print(res_df)
```
